```python
import numpy as np

a=np.array([
    [5,3,3],
    [4,3,2],
    [4,4,5]
],dtype=float)
```

```python
print(a)
```

```python
U,S,VT=np.linalg.svd(a)
```

converting the singular values into diagonal matrix

```python
print(S)
```

```python
sigma = np.diag(S)
```

Reconstruction of matrix

```python
a_reconstructed=U @ sigma @ VT
```

```python
print(a_reconstructed)
```

MOVIE_RATING_RECOMMENDATION

```python
import numpy as np

r=np.array([
    [5,3,np.nan],
    [4,np.nan,2],
    [np.nan,4,5]
],dtype=float)

r_org=r.copy()
```

```python
print(r_org)
```

```python
col_means=np.nanmean(r,axis=0)
```

```python
print(col_means)
```

```python
inds=np.where(np.isnan(r))
```

```python
r[inds]=np.take(col_means,inds[1])
```

```python
print(r)
```

RECONSTRUCTION OF MATRIX

```python
U,S,VT=np.linalg.svd(r)
```

```python
re_diag=np.diag(S)
```

```python
a_reconstructed=U @ re_diag @ VT
```

```python
print(a_reconstructed)
```

```python
k = 2

U_k = U[:, :k]
S_k = np.diag(S[:k])
VT_k = VT[:k, :]

R_pred = U_k @ S_k @ VT_k
```

```python
print(R_pred)
```

```python
print(r_org)
```

```python
missing_mask = np.isnan(r_org)
```

```python
predicted_values = R_pred[missing_mask]

print(predicted_values)
```
